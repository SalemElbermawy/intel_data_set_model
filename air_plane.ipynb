```python
number_cols=df.select_dtypes(include="number").columns
```

```python
numerical_data=df.loc[:,number_cols]
```

```python
matrix_corr=numerical_data.corr()
```

```python
plt.figure(figsize=(10,5))

sns.heatmap(matrix_corr,annot=True,fmt=".2f")

plt.show()
```

```python
matrix_corr.head()
```

```python
abs_corr_matrix=matrix_corr.abs()
```

```python
abs_corr_matrix.head()
```

```python
one_matrix=np.ones(shape=abs_corr_matrix.shape).astype(bool)
lower_trig=np.tril(one_matrix,k=-1)
lower_corr=abs_corr_matrix.where(lower_trig)
```

```python
lower_corr.head()
```

```python
lower_corr=lower_corr.unstack()
```

```python
lower_corr.head()
```

```python
type(lower_corr)
```

```python
lower_corr=lower_corr.dropna()
```

```python
sort_matrix=lower_corr.sort_values(ascending=False)
sort_matrix=sort_matrix.reset_index()
```
